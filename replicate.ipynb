```python
from google.colab import userdata
gemini_api_key = userdata.get('GEMINI_API_KEY')
openai_api_key = userdata.get('OPENAI_API_KEY')
```

```python
# 2. Import and set up LiteLLM
import os
from litellm import completion

# 3. API Keys (Set these as needed)
os.environ["GEMINI_API_KEY"]      = gemini_api_key   # OpenAI
os.environ["OPENAI_API_KEY"]      = openai_api_key      # Gemini, Google
# os.environ["REPLICATE_API_TOKEN"] = "r8-..."   # Llama on Replicate
```

```python
# 4. The unified function
def call_model(prompt, model="gpt-3.5-turbo", temperature=0.7, max_tokens=512):
    """Call any supported model by name."""
    msg = [{"role":"user", "content":prompt}]
    # If you get an error about missing API keys, check environment variables!
    # Explicitly pass API keys for respective models
    if model.startswith("gemini"):
        resp = completion(model=model, messages=msg, temperature=temperature, max_tokens=max_tokens)
    elif model.startswith("gpt"):
         resp = completion(model=model, messages=msg, temperature=temperature, max_tokens=max_tokens)
    else: # Add other models here as needed
        resp = completion(model=model, messages=msg, temperature=temperature, max_tokens=max_tokens)
    return resp.choices[0].message.content

# 5. Example calls
print("--- OpenAI GPT-3.5 ---")
print(call_model("Write a 2-line poem about llamas.", model="gpt-3.5-turbo"))

print("\n--- Gemini Pro ---")
print(call_model("Summarize the importance of water.", model="gemini/gemini-1.5-pro"))

# print("\n--- Llama on Replicate ---")
# print(call_model("List 3 uses for paperclips.", model="replicate/llama-2-70b-chat"))
```
